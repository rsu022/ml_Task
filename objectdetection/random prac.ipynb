```python
import random

a = [1,2,3,4,5,6]
print(random.choice(a)) #list ma bata random num choose garna

s = "Hello World"
print(random.choice(s)) #string ma bata random char choose garna

tup = (1,2,3,4,5)
print(random.choice(tup)) #tuple ma bata random num choose garna
```

sample() le multiple items pick garcha from sequences without repeating

```python
from random import sample

a = [1,2,3,4,5,6]
print(sample(a,3)) #list ma bata 3 ota random num choose garna

b = (4,5,6,7,8)
print(sample(b,2)) #tuple ma bata 2 ota random num choose garna

c = "45678"
print(sample(c,4)) #string ma bata 4 ota random char choose garna
```

```python
import random
a = [1,2,3,4,5,6]

random.shuffle(a) #list ko order random garna
print("After shuffle :")
print(a)

random.shuffle(a)
print("\nSecond shuffle :")
print(a)
```
