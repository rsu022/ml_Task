Object Detection with opencv and yolov8(ultralytics)

Object detection is a task in cv used to detect objects in an image or video. Opencv chai image /video processing and yolo - you only look once for real time detection

in yolo, we have to look only once in the network and pass forward once to make the final predictions

How it works?
1. Yo model le chai image accept garcha as a input.
Resize garcha image lai 448 * 448 pixels ma ensuring that the aspect ration chai preserve hos using padding. 

2. Image lai preprocess gare pachi chai the image is passed through deep CNN architecture jasma 24 convolutional layers and 4 max pooling layers huncha.

3. no. of parameters and compress channels reduce garna lai 1*1 convolutions use huncha ani they are followed by 3*3 convolutions to capture spatial patterns in feature maps.

YOLO:
first ma divide garch image lai grid ma.
each grid cell predicts:
bounding box: where the object might be x,y, width, height
objectness score: how confident the model is that there's something there
class probability: what kind of object it is 

sabai prediction combine huncha ani final detection output (boxes+labels+confidence)

```python
import cv2 
import random 
import numpy as np
import lap
from ultralytics import YOLO
from matplotlib import pyplot as plt
```

YOLOv8n.pt : nano (fastest, less accurate)
YOLOv8x.pt : Extra Large (most accurate tara slow)

```python
yolo = YOLO("yolov8n.pt")
```

```python
def getColours(cls_num):
    """Generate unique colors for each class ID"""
    random.seed(cls_num) #makes color consistent across frames
    return tuple(random.randint(0, 255) for _ in range(3))
```

```python
video_path = r"C:\Users\reshu\Desktop\ml_Task\objectdetection\video.mp4"
videoCap = cv2.VideoCapture(video_path)
```

Process Video and Detect Objects

videoCap.read() : Reads one frame from video
yolo.track(frame,stream=True): Runs Yolov8 object detection
result.boxes: contains bounding boxes for detected objects
box.conf[0]: Confidence score of detection
box.xyxy[0]: Coordinates of bounding box (x1,y1,x2,y2)
cv2.rectangle(): Draws bounding box
cv2.putText(): Adds class name + confidence on frame

```python
frame_count = 0

while True:
    ret, frame = videoCap.read()
    if not ret:
        break
    results = yolo.track(frame, stream=True)

    for result in results:
        class_names = result.names
        for box in result.boxes:
            if box.conf[0] > 0.4:
                x1, y1, x2, y2 = map(int, box.xyxy[0])

                cls = int(box.cls[0])
                class_name = class_names[cls]

                conf = float(box.conf[0])

                colour = getColours(cls)

                cv2.rectangle(frame, (x1, y1), (x2, y2), colour, 2)

                cv2.putText(frame, f"{class_name} {conf:.2f}",
                            (x1, max(y1 - 10, 20)), cv2.FONT_HERSHEY_SIMPLEX,
                            0.6, colour, 2)

    if frame_count < 20:
        cv2_imshow(frame)
    else:
        break

    frame_count += 1

videoCap.release()
```
