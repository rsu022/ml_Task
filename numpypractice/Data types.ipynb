Checking the Data Type of an Array

dtype le kun data type of array vancha.
used as arr.dtype

```python
arr = np.array([1,2,3,4])
print(arr.dtype)
```

```python
#Get the data type of an array containing strings:
#< → little-endian (how bytes are stored in memory, usually you don’t worry about it)
#U → Unicode string
#5 → maximum number of characters each string can have
#So <U5 = Unicode string of max length 5.

arr = np.array(['hello','world'])
print(arr.dtype)    
```

Creating Arrays With a Defined Data Type

array() function bhitra dtype use garera data type lekhna milcha j bhayeni

```python
#dtype='S' → tells NumPy to store all elements as bytes/strings, not numbers.

#So even though you wrote numbers [1, 2, 3, 4], they become byte-strings.
#|S1 → data type = string of length 1

arr = np.array([1, 2, 3, 4], dtype='S')
print(arr)
print(arr.dtype)
```

```python
#Create an array with data type 4 bytes integer:

arr = np.array([1, 2, 3, 4], dtype='i4')

print(arr)
print(arr.dtype)
```

Converting Data Type on Existing Arrays

astype()- changes the data type of an existing array by making a copy of that array

The data type can be specified using a string, like 'f' for float, 'i' for integer etc. or you can use the data type directly like float for float and int for integer.

```python
#Change data type from float to integer by using 'i' as parameter value:

arr = np.array([1.2,1.3,1.4])
eg = arr.astype('i')
print(eg)
print(eg.dtype)
```

```python
#Change data type from float to integer by using int as parameter value:
arr = np.array([1.2,1.3,1.4])
eg = arr.astype('int')
print(eg)
print(eg.dtype)
```

astype(bool) converts each element to a boolean:

0 → False

any non-zero number → True (negative or positive doesn’t matter)

```python
#From int to boolean

arr = np.array([1,2,3])
                     
eg=arr.astype('bool')
print(eg)
print(eg.dtype)
```

```python
arr = np.array([-1, 0, 1])
newarr = arr.astype(bool)
print(newarr)
```
