Dimensions in Array

0-D Array or Scalars vaneko elements in array.
Each value in an array is a 0 D array

```python
arr = np.array(43)
print(arr)
```

1D array has 0 d array as its elements and is called uni -dimensional as well

```python
arr = np.array([1,2,3,4,5]) #Create a 1-D array containing the values 1,2,3,4,5:
print(arr)
```

2D array has 1D elements and used to represent matrix or 2nd order tensors

Matrix operation ko lagi numpy.mat use huncha

[[ ]] - bracket used

```python
#Create a 2-D array containing two arrays with the values 1,2,3 and 4,5,6:
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr)
```

3-D array is An array that has 2-D arrays (matrices) as its elements 

These are often used to represent a 3rd order tensor.
[[[ ]]] -brackets used

```python
#Create a 3-D array with two 2-D arrays, both containing two arrays with the values 1,2,3 and 4,5,6:
arr = np.array([[[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]]])
print(arr)
```

Check Number of Dimensions?
ndim attribute is used to return an integer jasle kati oota dimensions cha array ma vanncha

```python
a = np.array(42)
b = np.array ([1,2,3,4])
c = np.array([[1, 2, 3], [4, 5, 6]])
d = np.array([[[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]]])

print(a.ndim)
print(b.ndim)
print(c.ndim)
print(d.ndim)
```

Higher Dimensional Arrays

array ma jati oota pani dimensions huna sakcha.
ndim le chai kati oota cha vanera vancha
ndimn le chai hamlai define garna dencha

```python
arr = np.array([1, 2, 3, 4], ndmin=2)
print(arr)
print('number of dimensions :', arr.ndim)
```
