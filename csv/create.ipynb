```python
import csv
import random #random marks haru generate garna
```

```python
with open("students_score.csv", "w", newline="") as file:  # file opens here
    writer = csv.writer(file)
    writer.writerow(["id", "name", "math", "science", "english"])
    
    for i in range(1, 101): # 1 to 100 students
        writer.writerow([ 
            i,               #id ko lagi
            f"Student_{i}", #student name ko lagi student_1 and so on
            random.randint(0, 100),
            random.randint(0, 100),
            random.randint(0, 100)
        ])
    print("File created!")
```

```python
import pandas as pd
```

```python
df = pd.read_csv("students_score.csv")
```

```python
df.head()
```

```python
#Avg score column
df['average_score'] = (df['math'] + df['science'] + df['english']) / 3 #(mean ko formula)
print(df.to_string())
```

```python
#find top 3 students with highest avg score
print("Top 3 students:")
print(df.sort_values('average_score', ascending=False).head(3)[['id','name','average_score']])
```

```python
#Group students into pass and fail and count them
df['status'] = df['average_score'].apply(lambda x: 'Pass' if x >= 40 else 'Fail')
print("Pass/Fail count:")
print(df['status'].value_counts())
```

```python
print(df.to_string())
```

```python
import numpy as np
```

```python
#convert subject score into numpy array
scores = df[['math','science','english']].to_numpy()
```

```python
#Calculate subject wise mean, min and max using numpy
print("Subject-wise Mean:", np.mean(scores, axis=0)) #axis = 0 vaneko column wise ho ani 1 vaneko row wise
print("Subject-wise Min:", np.min(scores, axis=0))
print("Subject-wise Max:", np.max(scores, axis=0))
```

```python
#Save avg score and status in new csv
df.to_csv("students_results.csv", index=False)
print("Saved to 'students_results.csv'")
```
