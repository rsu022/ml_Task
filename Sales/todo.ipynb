```python
import pandas as pd
import random
from datetime import datetime, timedelta
```

```python
products = ["Laptop", "Phone", "Tablet", "Headphones", "Monitor", "Keyboard", "Mouse", "Printer", "Camera", "Speaker"]
categories = ["Electronics", "Accessories"]

rows = []
start_date = datetime(2023, 1, 1)
```

```python
for i in range(1, 101):   # 100 rows
    order_id = i
    product = random.choice(products)
    category = "Electronics" if product in ["Laptop", "Phone", "Tablet", "Monitor", "Camera"] else "Accessories"
    price = round(random.uniform(50, 1500), 2)
    quantity = random.randint(1, 5)
    order_date = start_date + timedelta(days=random.randint(0, 365))
    rows.append([order_id, product, category, price, quantity, order_date.strftime("%Y-%m-%d")])
```

```python
df = pd.DataFrame(rows, columns=["order_id", "product_name", "category", "price", "quantity", "order_date"])
df.to_csv("sales_data.csv", index=False)
print("CSV file created: sales_data.csv")
```

Load the CSV

```python
df = pd.read_csv("sales_data.csv", parse_dates=["order_date"])
print(df.head()) 
```

```python
#Total amount column created
df["total_amount"] = df["price"] * df["quantity"]
```

```python
#Group By Category
category_summary = df.groupby("category").agg(
    total_sales=("total_amount", "sum"),
    avg_sales_per_order=("total_amount", "mean"),
    num_orders=("order_id", "count")
).reset_index()

print(category_summary)
```

```python
#Top 5products by total sales
top_products = df.groupby("product_name")["total_amount"].sum().nlargest(5)
print("Top 5 products by sales:\n", top_products)
```

```python
import numpy as np
```

```python
# Standard deviation of total_amount
std_total = np.std(df["total_amount"])
print("Std Dev of total_amount:", std_total)

```

```python
# Percentiles
p25, p50, p75 = np.percentile(df["total_amount"], [25, 50, 75])
print("25th:", p25, "50th:", p50, "75th:", p75)
```

```python
# Normalize price column (min-max scaling)
df["price_normalized"] = (df["price"] - df["price"].min()) / (df["price"].max() - df["price"].min())
```

```python
#Extract Month-Year + Monthly Sales
df["month"] = df["order_date"].dt.month
df["year"] = df["order_date"].dt.year

monthly_sales = df.groupby(["year", "month"])["total_amount"].sum().reset_index()
print(monthly_sales)
```

```python
#Find Months Below Overall Average
overall_avg = df["total_amount"].mean()

below_avg_months = monthly_sales[monthly_sales["total_amount"] < overall_avg]
print("Months below average:\n", below_avg_months)
```

```python
#Save Outputs
category_summary.to_csv("category_summary.csv", index=False)
df.to_csv("sales_results.csv", index=False)

print("Files saved: category_summary.csv & sales_results.csv")
```
